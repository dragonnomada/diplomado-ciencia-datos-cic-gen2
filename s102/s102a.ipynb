```python
# Ejemplo E401.1 - Determinar el área de un triángulo

# Definimos las variables para retener el valor de los lados del triángulo
l1 = 3
l2 = 4
l3 = 5

s = (l1 + l2 + l3) / 2 # calculamos la mediatriz

a = (s * (s - l1) * (s - l2) * (s - l3)) ** 0.5 # Raíz cuadrada -> ^(1/2)

print(f"El área del triángulo <{l1}, {l2}, {l3}> es {a:.2f}")
```

```python
# Ejemplo E401.2 - Determinar si un punto está dentro de un circulo

px = 3 # punto en x
py = 5 # punto en y

cx = 1 # centro del círculo en x
cy = 2 # centro del círculo en y
cr = 4 # radio

# El punto se aleja del centro del círculo en la distancia `d`
# y si la distancia supera el radio, entonces el punto ya no puede estar dentro del círculo
# por lo tanto, si `d <= r`, entonces, el punto está dentro del círculo

dx = px - cx
dy = py - cy

d = (dx ** 2 + dy ** 2) ** 0.5 # Distancia euclidea con Teorema de Pitágoras (C^2 = A^2 + B^2)
# d^2 = dx^2 + dy^2

esta_en_circulo = d <= cr

esta_en_circulo
```

```python
# Ejemplo E401.2b - Determinar si un punto está dentro de un circulo

px = float(input("Punto X: ")) # punto en x
py = float(input("Punto Y: ")) # punto en y

cx = float(input("Circulo X: ")) # centro del círculo en x
cy = float(input("Circulo Y: ")) # centro del círculo en y
cr = float(input("Circulo Radio: ")) # radio

# El punto se aleja del centro del círculo en la distancia `d`
# y si la distancia supera el radio, entonces el punto ya no puede estar dentro del círculo
# por lo tanto, si `d <= r`, entonces, el punto está dentro del círculo

dx = px - cx
dy = py - cy

d = (dx ** 2 + dy ** 2) ** 0.5 # Distancia euclidea con Teorema de Pitágoras (C^2 = A^2 + B^2)
# d^2 = dx^2 + dy^2

esta_en_circulo = d <= cr

repuestas = {True: 'Sí', False: 'No'}
print(f"Punto (x={px}, y={py}) -> Circulo (x={cx}, y={cy} | radio={cr}) -> Está en circulo? {repuestas[esta_en_circulo]}")
```

```python
# Estructura de control: Un bloque asociado y controlado por una condición o un patrón estructurado

# Condicional: SI <condición> ENTONCES <bloque>

# SINTAXIS: <estructura>: <bloque>
# 
# <bloque> - Son sentencias anidadas bajo la regla de indentanción

if 5 > 3:
    print("5 es mayor a 3")
    print("Esta línea también es parte del bloque")
    print("Esta otra línea también es parte del bloque")
print("Está última línea ya no es parte del bloque, sino del flujo principal")
```

```python
# Ejemplo E401.3 - Determinar si un triángulo es válido

# Tenemos los lados l1, l2 y l3 y es válido si:
# - l1 + l2 >= l3
# - l1 + l3 >= l2
# - l2 + l3 >= l1

# /    \
# ------

l1 = 1
l2 = 2
l3 = 5

if l1 + l2 >= l3:
    if l1 + l3 >= l2:
        if l2 + l3 >= l1:
            print("Triángulo es válido")
```

```python
# Ejemplo E401.3b - Determinar si un triángulo es válido

l1 = 1
l2 = 2
l3 = 5

if l1 + l2 >= l3:
    if l1 + l3 >= l2:
        if l2 + l3 >= l1:
            print("Triángulo es válido")
        else:
            print(f"Los lados (l2={l2} + l3={l3}) < l1={l1}")
    else:
        print(f"Los lados (l1={l1} + l3={l3}) < l2={l2}")
else:
    print(f"Los lados (l1={l1} + l2={l2}) < l3={l3}")
```

```python
# Secuenciadores/Iteradores: Es una estructura de control que permite repetir un bloque por cada elemento en una secuencia
# PARA <elemento> EN <secuencia>: <bloque>

for fruta in ["manzana", "kiwi", "melón"]:
    print(fruta)
```

```python
for indice, fruta in enumerate(["manzana", "kiwi", "melón"]):
    print(f"{indice}: {fruta}")
```

```python
for i in range(1, 1000): # Secuencia genera por un rango
    if i % 100 == 0: # condicional dentro del secuenciador
        print(i)
```

```python
modelos = ["robot_a", "robot_b"]
objetivos = ["limpiar", "cocinar", "barrer"]

for i, modelo in enumerate(modelos):
    for j, objetivo in enumerate(objetivos):
        print(f"({i}, {j}) {modelo} | {objetivo}")
```

```python
# Una función automatiza cálculos

import pandas

def cuantiles(serie, n = 4): # 4 cuantiles -> Cuartiles | 5 cuantiles -> Quintiles | ... | 10 cuantiles -> Deciles | ...
    serie = pandas.Series(serie)
    probas = []
    quantiles = []
    for i in range(n + 1): # i = 0, 1, 2, ..., n - 1, n
        p = i / n # n = 4 | 0, 0.25, 0.5, 0.75, 1
        q = serie.quantile(p)
        probas.append(p)
        quantiles.append(q)
    return pandas.Series(quantiles, index=probas)
```

```python
cuantiles([35, 27, 28, 65, 54, 23, 22, 56, 45, 41, 32, 37, 56, 54, 39, 45, 42])
```

```python
cuantiles([35, 27, 28, 65, 54, 23, 22, 56, 45, 41, 32, 37, 56, 54, 39, 45, 42], n=5)
```

```python
import seaborn

q = cuantiles([35, 27, 28, 65, 54, 23, 22, 56, 45, 41, 32, 37, 56, 54, 39, 45, 42], n=5)

seaborn.lineplot(x=q.index, y=q, marker="o", linestyle="--", markerfacecolor="red", color="gray")
```
